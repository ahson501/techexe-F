```python
#1️⃣ Import Necessary Libraries
#Why these libraries?
#1 pandas: For data handling
#2 numpy: For numerical computations
#3 train_test_split: To split the dataset into training and testing
#4 RandomForestRegressor: For LCST prediction (regression task)
#5 RandomForestClassifier: For Phase Behavior prediction (classification task)
#6 mean_squared_error: To evaluate LCST prediction

import pandas as pd
import os
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_squared_error, accuracy_score
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler   # For feature scaling
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import joblib     # For saving/loading models

from IPython.display import FileLink  # For downloading model


#2️⃣ Set the Working Directory and Load Data

# Set the working directory (if needed)
#e.g, Define the file path
# data_path = r"C:\Users\Sajeed\Desktop\polymer\data.csv"

print(os.getcwd())                                #✅to find the current directory. also use (%pwd)
print(os.listdir())                               #✅to list files & folders. also (%ls)

data = pd.read_csv("data.csv")                    #✅Load the CSV file into a Pandas DataFrame

print(data.head())           #✅Display the first few rows of the dataset to check the data is correctly loaded





#3️⃣Load and Define Features (X) and Targets (y)
# Define features (X) and target variable (y)

X_lcst = data[["blockA", "BlockB", "hydrophobic", "Temprature", "pH"]]
y_lcst = data["LCST"]

#Features (X_phase): Same as LCST features
#Target (y_phase): "Phase_Behavior"
X_phase = data[["blockA", "BlockB", "hydrophobic", "Temprature", "pH"]]
y_phase = data["Phase_Behavior"]

new_copolymer = [[7500, 3000, 55, 33, 6]]





#4️⃣Split the Data into Training and Testing Sets

X_lcst_train, X_lcst_test, y_lcst_train, y_lcst_test = train_test_split(X_lcst, y_lcst, test_size=0.2, random_state=42)

X_phase_train, X_phase_test, y_phase_train, y_phase_test = train_test_split(X_phase, y_phase, test_size=0.2, random_state=42)

#Why do we split data?
# Training Set (80%): Used to train the model
#Testing Set (20%): Used to check model accuracy


#5️⃣Train a Machine Learning Model
# Initialize and train the model 

#Train the Random Forest Models

#LCST Prediction - Random Forest Regressor
#Used for predicting continuous values (like temperature).
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42)  #✅Initialize the model
rf_regressor.fit(X_lcst_train, y_lcst_train)                            #✅ Train Model

#Used for categorical classification (Micelle, Vesicle, Lamella, etc.).
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_phase_train, y_phase_train)
 
    
    
    
    
#6️⃣Make Predictions on Test Data
# Make predictions on the test set 

# Predict LCST
y_lcst_pred = rf_regressor.predict(X_lcst_test)       # ✅ Predictions
print("Predicted LCST:", y_lcst_pred)                 # ✅ Output Results

# Predict Phase Behavior
#Since rf_classifier is a classification model, it will output a category (e.g., Micelle, Vesicle, Lamella).
new_phase_pred = rf_classifier.predict(new_copolymer)
print(f"Predicted Phase Behavior: {new_phase_pred[0]}")

#Since rf_regressor is a regression model, it will output a numerical LCST value.
new_lcst_pred = rf_regressor.predict(new_copolymer)
print(f"Predicted LCST: {new_lcst_pred[0]:.2f}")

y_phase_pred = rf_classifier.predict(X_phase_test)
print("Predicted Phase Behavior:", y_phase_pred)




#7️⃣Evaluate the Model
# Calculate accuracy
lcst_mse = mean_squared_error(y_lcst_test, y_lcst_pred)
print("Mean Squared Error (LCST Prediction):", lcst_mse)

#Why MSE?
# MSE measures how far predictions are from actual values.
#Lower MSE = Better model accuracy.




 #8️⃣Save the Trained Model in Your ML Folder
import joblib
# ✅ : Save the Trained Models
model_lcst_path = "/tf/notebooks/ML/lcst_model.pkl"
model_phase_path = "/tf/notebooks/ML/phase_model.pkl"


# Save models 
joblib.dump(rf_regressor, model_lcst_path)
joblib.dump(rf_classifier, model_phase_path)


print(f"\nModels saved successfully at:\n- {model_lcst_path}\n- {model_phase_path}")

 #9️⃣Load the Model & Make Predictions on New Data
# Load the saved models
loaded_lcst_model = joblib.load(model_lcst_path)
loaded_phase_model = joblib.load(model_phase_path)

# Example new data point (Must match feature structure)
new_copolymer = [[7500, 3000, 55, 33, 6]]  # Modify values as needed

# Predict LCST
predicted_lcst = loaded_lcst_model.predict(new_copolymer)
print(f"\nPredicted LCST: {predicted_lcst[0]:.2f}")

# Predict Phase Behavior
predicted_phase = loaded_phase_model.predict(new_copolymer)
print(f"\nPredicted Phase Behavior: {predicted_phase[0]}")


# Step 🔟: Download the Model (If Needed)
print("\nClick the link below to download the trained LCST model:")
FileLink(model_lcst_path)

```
