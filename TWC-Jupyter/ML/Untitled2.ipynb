```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_squared_error, accuracy_score
from sklearn.preprocessing import LabelEncoder

# ✅ Step 1: Define the Dataset
data = pd.DataFrame({
    "Block_A_MW": [5000, 7000, 6000, 8000, 9000],
    "Block_B_MW": [2000, 3000, 4000, 2000, 5000],
    "Hydrophobic_Fraction": [60, 50, 40, 70, 45],
    "Temperature": [25, 30, 35, 37, 40],
    "pH": [7, 6, 7, 5, 8],
    "LCST": [50, 60, 70, 45, 80],
    "Phase_Behavior": ["Micelle", "Vesicle", "Lamella", "Micelle", "Vesicle"]
})

# ✅ Step 2: Define Features & Targets
X_lcst = data[["Block_A_MW", "Block_B_MW", "Hydrophobic_Fraction", "Temperature", "pH"]]
y_lcst = data["LCST"]  # Regression target

X_phase = data[["Block_A_MW", "Block_B_MW", "Hydrophobic_Fraction", "Temperature", "pH"]]

# ✅ Encode Phase_Behavior Labels
le = LabelEncoder()
y_phase = le.fit_transform(data["Phase_Behavior"])  # Convert categories to numbers

# ✅ Step 3: Train-Test Split
X_lcst_train, X_lcst_test, y_lcst_train, y_lcst_test = train_test_split(X_lcst, y_lcst, test_size=0.2, random_state=42)
X_phase_train, X_phase_test, y_phase_train, y_phase_test = train_test_split(X_phase, y_phase, test_size=0.2, random_state=42)

# ✅ Step 4: Train Random Forest Models
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42)
rf_regressor.fit(X_lcst_train, y_lcst_train)

rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_phase_train, y_phase_train)

# ✅ Step 5: Predictions
y_lcst_pred = rf_regressor.predict(X_lcst_test)
y_phase_pred = rf_classifier.predict(X_phase_test)

# ✅ Decode Predicted Labels for Phase Behavior
y_phase_pred_labels = le.inverse_transform(y_phase_pred)

# ✅ Step 6: Evaluation
lcst_mse = mean_squared_error(y_lcst_test, y_lcst_pred)
phase_accuracy = accuracy_score(y_phase_test, y_phase_pred)

# ✅ Output Results
print("Predicted LCST:", y_lcst_pred)
print("Mean Squared Error (LCST Prediction):", lcst_mse)

print("Predicted Phase Behavior:", y_phase_pred_labels)
print("Accuracy (Phase Behavior Classification):", phase_accuracy)
```
