```python
# ✅ Step 1️⃣: Import Required Libraries
import os
import pandas as pd
import numpy as np
import joblib  # For saving/loading models
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler  # For feature scaling
from IPython.display import FileLink  # For downloading model

# ✅ Step 2️⃣: Print Current Working Directory
print("Current Working Directory:", os.getcwd())  # Should print: /tf/notebooks/ML

# ✅ Step 3️⃣: Load the Dataset
data_path = "/tf/notebooks/ML/data.csv"
data = pd.read_csv(data_path)
print("\nDataset Loaded Successfully!")

# ✅ Step 4️⃣: Define Features (X) and Targets (y)
# Features for both models
features = ["blockA", "BlockB", "hydrophobic", "Temprature", "pH"]

# LCST Regression
X_lcst = data[["blockA", "BlockB", "hydrophobic", "Temprature", "pH"]]
y_lcst = data["LCST"]

# Phase Behavior Classification
X_phase = data[["blockA", "BlockB", "hydrophobic", "Temprature", "pH"]]
y_phase = data["Phase_Behavior"]

# ✅ Step 5️⃣: Split Data into Training & Testing Sets
X_lcst_train, X_lcst_test, y_lcst_train, y_lcst_test = train_test_split(X_lcst, y_lcst, test_size=0.2, random_state=42)
X_phase_train, X_phase_test, y_phase_train, y_phase_test = train_test_split(X_phase, y_phase, test_size=0.2, random_state=42)

# ✅ Step 6️⃣: Feature Scaling (Important for Stability)
scaler = StandardScaler()
X_lcst_train_scaled = scaler.fit_transform(X_lcst_train)
X_lcst_test_scaled = scaler.transform(X_lcst_test)
X_phase_train_scaled = scaler.fit_transform(X_phase_train)
X_phase_test_scaled = scaler.transform(X_phase_test)

# ✅ Step 7️⃣: Train the Models
# Regression Model (LCST Prediction)
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42)
rf_regressor.fit(X_lcst_train_scaled, y_lcst_train)

# Classification Model (Phase Behavior Prediction)
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_phase_train_scaled, y_phase_train)

print("\nModels Trained Successfully!")

# ✅ Step 8️⃣: Save the Trained Models
model_lcst_path = "/tf/notebooks/ML/lcst_model.pkl"
model_phase_path = "/tf/notebooks/ML/phase_model.pkl"
scaler_path = "/tf/notebooks/ML/scaler.pkl"

# Save models and scaler
joblib.dump(rf_regressor, model_lcst_path)
joblib.dump(rf_classifier, model_phase_path)
joblib.dump(scaler, scaler_path)

print(f"\nModels saved successfully at:\n- {model_lcst_path}\n- {model_phase_path}")

# ✅ Step 9️⃣: Load the Models & Make Predictions on New Data
# Load the saved models
loaded_lcst_model = joblib.load(model_lcst_path)
loaded_phase_model = joblib.load(model_phase_path)
loaded_scaler = joblib.load(scaler_path)

# Example new data point (Must match feature structure)
new_copolymer = [[7500, 3000, 55, 33, 6]]  # Modify values as needed

# Scale the new data point
new_copolymer_scaled = loaded_scaler.transform(new_copolymer)

# Predict LCST
predicted_lcst = loaded_lcst_model.predict(new_copolymer_scaled)
print(f"\nPredicted LCST: {predicted_lcst[0]:.2f}")

# Predict Phase Behavior
predicted_phase = loaded_phase_model.predict(new_copolymer_scaled)
print(f"\nPredicted Phase Behavior: {predicted_phase[0]}")

# ✅ Step 🔟: Download the Model (If Needed)
print("\nClick the link below to download the trained LCST model:")
FileLink(model_lcst_path)


#Would you like to:

#Optimize Hyperparameters? (GridSearchCV)
#Add Model Evaluation Metrics? (Accuracy, R² Score, etc.)
#Visualize Feature Importance? (Matplotlib & Seaborn)
```
